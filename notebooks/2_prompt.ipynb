###Static prompts

```python
from langchain_openai import ChatOpenAI
llm = ChatOpenAI(model= "gpt-5-mini")
prompt = [("system", "You are a python developer"),
("user", "Explain how to sort an array")]
res = llm.invoke(prompt)
print(res.content)
```

###Dynamic prompts

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
llm = ChatGoogleGenerativeAI(model= "gemini-3.5-flash-lite")
prompts = ChatPromptTemplate.from_messages([
    ("system", "You are a translator and translate into {language}"),
    ("user", "{query}")
])
final_prompt = prompts.format_prompt(language="Hindi", query="I want a job.")
res = llm.invoke(final_prompt)
print(res.text)
```

###Chains

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
llm = ChatGoogleGenerativeAI(model= "gemini-3.5-flash-lite")
out = StrOutputParser()
def tranform_case(result: str):
    return result.upper()
prompts = ChatPromptTemplate.from_messages([
    ("system", "You are a translator and translate into {language}"),
    ("user", "{query}")
])
chains = prompts | llm | out | tranform_case
res = chains.invoke({"language": "Hinglish", "query": "I want a good job."})
print(res)
```
