#### How to create a tool using python function

```python
from dotenv import load_dotenv
load_dotenv()
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.tools import tool
from langchain.agents import create_agent
llm = ChatGoogleGenerativeAI(model = "gemini-3.5-flash-lite")
```

```python
@tool
def add_numbers(a : int, b : int) -> int:
    """Add two numbers together.
    
    Args:
        a (int): The first number.
        b (int): The second number.
        
    Returns:
        int: The sum of the two numbers.
    """
    return a + b
```

```python
add_numbers.invoke({"a": 5, "b": 10})
```

```python
@tool
def multiply_numbers(a : int, b : int) -> int:
    """Product of two numbers.
    
    Args:
        a (int): The first number.
        b (int): The second number.
        
    Returns:
        int: The product of the two numbers.
    """
    return a * b
```

#### how to create an agent that connect LLM and Tools

```python
agent = create_agent(
    model = llm,
    tools = [add_numbers, multiply_numbers],
    system_prompt = (
        "You are a math teacher and use the available tools for calculations."
    )
)

response = agent.invoke({
    "messages": [
        {"role": "user",
        "content": "What is the sum of 2 + 30 * 10?"}
    ]
})
```

```python
print(response["messages"][-1].text)
```
