#### Using Gemini

```python
from dotenv import load_dotenv
load_dotenv()
from langchain_google_genai import ChatGoogleGenerativeAI
llm = ChatGoogleGenerativeAI(model = "gemini-3.5-flash-lite", streaming = True)
```

```python
question = "What is GenAi?"
response = llm.stream(question)

for chunk in response:
    print(chunk.text, end="")
```

#### Using groq

```python
from langchain_groq import ChatGroq
llm2 = ChatGroq(model = "openai/gpt-oss-20b", streaming = True)

response2 = llm2.stream(question)
for chunk in response2:
    print(chunk.text, end="")
```
